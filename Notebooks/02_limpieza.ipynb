## 3. Limpieza de `bank`

### 3.1 Eliminación de columnas

- `Unnamed: 0`: No nos aporta información, asi que no sirve de identificador (para eso está la columna `id_`)
- `latitude` y `longitude`: Al ser coodenadas de Estados Unidos me parecen irrelevantes al ser un banco portugués.

```python
bank = bank.drop(columns=["Unnamed: 0", "latitude", "longitude"])
bank.head()
```

### 3.2 Conversión de datos
`cons.price.idx`, `cons.conf.idx`, `euribor3m` y `nr.employed` son columnas numéricas pero con el `.info()` vemos que son str, lo cambiamos


```python
bank.info()
```

```python
def texto_a_numero(columnas):
    """Convierte una columna str a float"""

    return pd.to_numeric(columnas.str.replace("," , "."))

columnas_texto_numero = ["cons.price.idx", "cons.conf.idx", "euribor3m", "nr.employed"]

for c in columnas_texto_numero:
    bank[c] = texto_a_numero(bank[c])

bank[columnas_texto_numero].info()
```

### 3.3 Fechas: `date`, `contact_month` y `contact_year`
`date` es un texto con el mes escrito en español (por ejemplo, `2-agosto-2019`), que pandas no puede interpretar directamente. Uso un diccionario con los meses para convertirla a fecha con una función aplicada a la columna.

Las columnas `contact_month` y `contact_year` aparecían en el enunciado pero **no existían en el archivo**, así que las creo a partir de la fecha. Las 248 filas sin fecha quedan sin mes ni año (no invento un dato que no tengo).

```python
MESES = {
    "enero": 1, "febrero": 2, "marzo": 3, "abril": 4, "mayo": 5, "junio": 6,
    "julio": 7, "agosto": 8, "septiembre": 9, "octubre": 10, "noviembre": 11, "diciembre": 12,
}


def convertir_fecha(texto):
    """Convierte '2-agosto-2019' en una fecha. Si no hay dato, devuelve NaT."""
    if pd.isna(texto):
        return pd.NaT
    dia, mes, anio = texto.split("-")
    return pd.Timestamp(year=int(anio), month=MESES[mes], day=int(dia))


bank["date"] = bank["date"].apply(convertir_fecha)
bank["contact_month"] = bank["date"].dt.month.astype("Int64")
bank["contact_year"] = bank["date"].dt.year.astype("Int64")

bank[["date", "contact_month", "contact_year"]].head()
```

### 3.4 Variables de texto: unificar formato y tratar nulos
- **Formato:** `marital` y `poutcome` estaban escritas en mayúsculas y el resto en minúsculas. Paso todas las columnas de texto a minúsculas y quito espacios sobrantes.
- **Nulos en `job` (0,8 %), `marital` (0,2 %) y `education` (4,2 %):** son columnas de categorías y los porcentajes son bajos. No borro filas, porque perdería información útil de otras columnas, ni invento una profesión o un estado civil. Creo una opción para cambiar NaN a `desconocido`.

```python
columnas_texto_clean = ["job", "marital", "education", "contact", "poutcome", "y"]

for col in columnas_texto_clean:
    bank[col] = bank[col].str.strip().str.lower()

bank[columnas_texto_clean].head()
```

```python
columnas_con_nulos = ["job", "marital", "education"]
bank[columnas_con_nulos] = bank[columnas_con_nulos].fillna("desconocido")

for col in columnas_texto_clean:
    print(col, bank[col].unique().tolist())
```

```python
bank[columnas_con_nulos].isna().sum()
```

### 3.5 Variables 0/1
- **`default`, `housing` y `loan`** valen 0 o 1, pero pandas las cargó como decimales (`0.0`, `1.0`) por tener nulos. Las paso a enteros con soporte para nulos (`Int64`).
- **`default` se mantiene**, aunque casi todos sus valores sean 0 (solo 3 clientes tienen un 1). Valoraré en el análisis si aporta algo.
- **`y`** (si el cliente suscribió el depósito) está en texto. Creo `suscrito` (`Int64`), con 1 para `yes` y 0 para `no`, que facilitará calcular tasas de contratación y correlaciones.

```python
columnas_si_no = ["default", "housing", "loan"]
bank[columnas_si_no] = bank[columnas_si_no].astype("Int64")


bank["suscrito"] = (bank["y"] == "yes").astype(int)

bank[columnas_si_no].dtypes
```

```python
bank.info()
```

### 3.6 Nulos en indicadores económicos: `cons.price.idx` y `euribor3m`
Estos indicadores dependen del periodo económico, por lo que rellenamos los nulos agrupando por `emp.var.rate`, `nr.employed` y `cons.conf.idx`.

- `cons.price.idx`: usamos la mediana del grupo, ya que tiene un único valor por periodo.
- `euribor3m`: usamos también la mediana del grupo para conservar las diferencias entre periodos.

```python
claves_periodo = ["emp.var.rate", "nr.employed", "cons.conf.idx"]

for col in ["cons.price.idx", "euribor3m"]:
    mediana_grupo = bank.groupby(claves_periodo)[col].transform("median")
    bank[col] = bank[col].fillna(mediana_grupo)

bank[["cons.price.idx", "euribor3m"]].isna().sum()
```

### 3.7 Nulos en `age`
Tiene un 11,9 % de nulos y ninguna otra columna permite deducir la edad. Prefiero no borrar tantas filas, así que la relleno con la **mediana**, creo otra columna `age_rellenada`

```python
bank["age_rellenada"] = bank["age"].isna()

bank["age"] = bank ["age"].fillna(bank["age"].median())
bank["age"] = bank["age"].astype(int)

bank["age_rellenada"].sum()
```

### 3.8 `pdays`: el valor 999

El valor `999` representa que el cliente no había sido contactado anteriormente, no un número real de días.

- Sustituyo `999` por nulos para evitar alterar los cálculos.
- Creo `contactado_antes`, que indica si hubo contactos previos según `previous`.
- Mantengo sin modificar los casos inconsistentes (`previous > 0` y `pdays = 999`), ya que no conocemos el dato correcto.

```python
bank["contactado_antes"] = (bank["previous"] > 0).astype(int)
bank["pdays"] = bank["pdays"].replace(999, np.nan)

bank[["pdays", "previous", "contactado_antes"]].describe()
```

### 3.9 `duration` igual a 0
Una llamada de 0 segundos significa que **no llegó a establecerse**, así que no es una interacción real con el cliente. Son muy pocas filas, por lo que las elimino.

```python
filas_antes = len(bank)
bank = bank[bank["duration"] > 0].reset_index(drop=True)
print("Filas eliminadas:", filas_antes - len(bank))
```

### 3.10 Estado de `bank` tras la limpieza
Los nulos que quedan son **intencionados**: `default`, `housing` y `loan` (valor desconocido), `date`, `contact_month` y `contact_year` (sin fecha) y `pdays` (sin contacto previo).

```python
bank.isna().sum()
```

## 4. Limpieza de `customers`

### 4.1 Unir las tres hojas
Las tres hojas tienen **las mismas columnas** y lo que quiero es una tabla más larga, así que las apilo con `concat`. Uso `ignore_index=True` para que el índice se numere de nuevo, de 0 a 43.169, ya que cada hoja conserva el suyo y los números se repetirían.

```python
customer_concat = pd.concat([customer_2012, customer_2013, customer_2014], axis = 0, join= "outer", ignore_index= True)
customer_concat.head()
```

```python
customer_concat.shape
```

```python
customer_concat.info()
```

### 4.2 Columnas sobrantes y comprobaciones
- Elimino `Unnamed: 0`, que es el mismo índice sobrante que en `bank`.
- Compruebo que el identificador `ID` no se repite entre hojas (un cliente no puede aparecer dos veces) y que no hay nulos (esto ya lo revise en la carga y exploración)
- `Dt_Customer` ya se carga como fecha, así que no necesita conversión.

```python
customers = customer_concat.drop(columns="Unnamed: 0")

print("ID repetidos:", customers["ID"].duplicated().sum())
print("Nulos en total:", customers.isna().sum().sum())
print(customers.dtypes)
```

```python
customers.info()
```

## 5. Unir `bank` y `customers`
Cruzo las dos tablas con `merge`. La clave se llama distinto en cada una (`id_` en `bank`, `ID` en `customers`), así que la indico con `left_on` y `right_on`.

```python
bank_customers = pd.merge(bank, customers, left_on="id_", right_on="ID", how="left")
bank_customers = bank_customers.drop(columns="ID")
bank_customers.shape
```
